```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Make plots appear inside the notebook
%matplotlib inline
```

```python
df = pd.read_csv("../data/raw/diabetes.csv")

df.head()
```

```python
df.shape
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Outcome", data=df)

plt.title("Distribution of Diabetes Cases")

plt.show()
```

```python
df.hist(figsize=(15,10), bins=20)

plt.suptitle("Distribution of All Features", fontsize=16)

plt.tight_layout()

plt.show()
```

```python
clean_df = df.copy()
```

```python
columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for column in columns:
    median = clean_df[column].replace(0, float("nan")).median()
    clean_df[column] = clean_df[column].replace(0, median)
```

```python
for column in columns:
    print(column, ":", (clean_df[column] == 0).sum())
```

```python
clean_df.to_csv(
    "../data/processed/diabetes_cleaned.csv",
    index=False
)
```
